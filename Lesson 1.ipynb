# Section 1: Shift cipher

There are a few words we need to learn first to understand the examples. Within cryptography, people send secret messages to each other. We call the messages *plaintext*. Remember that the plaintext is what humans can understand and read. As we only want the reciever of the message to understand what it says, we need to turn this message into *ciphertext*. This ciphertext does not make any sense so anyone who intercepts the message will not be able to understand it. To turn the plaintext into ciphertext, we need to *encrypt* the message. This encryption will be our secret key that only two people will now. Then to turn the ciphertext into plaintext, the reciever needs to *decrypt* the message. If this does not make any sense, then do not worry. The example below will help you understand. 

### Example 1: Shift cipher of 1 using English

Imagine we have the english alphabet,

```python
#| echo: false
#| label: alphabet
#| fig-cap: "The English Alphabet"
from IPython.display import Image, display
display(Image(filename='Alphabet.png'))
```

I want to use the alphabet to code a secret message. Alice wants to send the word "Svitlo" to Bob. Alice and Bob have agreed to use a shift cipher of 1. This means that each letter in the plaintext (Svitlo) will move one letter to the right. This can be seen in the diagram below:

```python
#| echo: false
#| label: alphabet-shifted
#| fig-cap: "Diagram explaining a shift cipher of value one."
from IPython.display import Image, display
display(Image(filename='Shift cipher 1.png'))
```

For "Svitlo", the S will become a T, the V will become a W and so on. Meaning we get: $$\text{``Svitlo'' in plaintext}=\text{``Twjump'' in ciphertext}$$

So Alice will send "Twjump" to Bob. Bob knows that Alice has shifted every letter one place to the right so to decrypt the message Bob needs to shift every letter one to the left. By doing this Bob, will successfully decrypt Alices message "Svitlo".

Earlier we learnt 4 words:
- Plaintext: The message you want to send someone
- Ciphertext: A string of letters that do not make sense.
- Encrypt: Use a key to turn the plaintext into ciphertext.
- Decrypt: Use a key to turn the ciphertext into plaintext.

We can demonstrate the relationship between these 4 words using the diagram below.

```python
#| echo: false
#| label: plaintext-ciphertext
#| fig-cap: "Diagram explaining plaintext, ciphertext, encryption and decryption."
from IPython.display import Image, display
display(Image(filename='Plaintext ciphertext explanation.png'))
```

# Section 2: Modular arithmetic
This next section is important for studying cryptography, so if you have never seen this before, I encourage you to take your time to make sure you understand this well.

Modular arithmetic is sometimes called "clock arithmetic" because of how we can use a clock to explain it. Imagine we have a 12 hour clock, where the 12 has been replaced by a zero. It would look something like this:

```python
#| echo: false
#| label: clock-mod-example
from IPython.display import Image, display
display(Image(filename='Clock mod example.png', width=350,height=350))
```

We can see that this clock tells us that the time is 2 o'clock. If wait 8 hours, then the clock will show 10 o'clock, which can be seen below:

```python
#| echo: false
#| label: clock-mod-example-2
from IPython.display import Image, display
display(Image(filename='Clock mod example 2.png', width=800,height=350))
```

This is easy to understand, as $2+8=10$. But what happens if we wait another 8 hours?

#| echo: false
#| label: clock-mod-example-3
from IPython.display import Image, display
display(Image(filename='Clock mod example 3.png', height=350, width=800))

As you can see, waiting another 8 hours means it will go to 6 o'clock. On this clock we cannot have any number which is 12 or greater. This means when we go past the number 11, we have to take away 12 until we get a number between zero and eleven inclusive.
So when we add 8 hours the seconds time what we are doing is: $10+8=18$, but because we know that $18$ is 12 or greater, then we have to take away 12. When we take away 12 we get $18-12=6$, which is why the clock shows 6 o'clock. We will now look at a spinner which only has 4 numbers.

```python
#| echo: false
#| label: spinner-mod-example
from IPython.display import Image, display
display(Image(filename='Spinner mod example.png'))
```

We can see that we now have a spinner with only 4 numbers (or if you want then you can imagine it to be some sort of Alien clock). The spinner is currently showing the number 2. If we moved the hand on the spinner 3 times clockwise, then it would end up on the number 1. This can be shown below:

```python
#| echo: false
#| label: spinner-mod-example-2
from IPython.display import Image, display
display(Image(filename='Spinner mod example 2.png', height=350, width=800))
```

In this case, the calculation we are doing is $2+3=5$. However the clock does not show the number $5$ meaning we need to take away $4$ until we get a number which does appear on the clock. Hence, we do $5-4=1$ so the clock shows the number $1$.

Now we need to translate the clock and spinner into Mathematics. When a clock has 12 numbers, we say that we are doing the calculations$\mod 12$. Writing out the example with the clock Mathematically would look like this: $$2+8=10 \text{ (mod } 12)$$ $$10+8=18 \equiv 6 \text{ (mod } 12)$$

In the second example with the spinner (which has 4 numbers), we would write: $$2+3=5 \equiv 1\text{ (mod }4)$$

Now before some practice questions, we will look at some examples.

### Example 1
Calculate $13 + 21 \text{ (mod 26)}$:

First, we will calculate $13+21$, which will give $34$. $34$ is greater than 26 meaning we need to take away 26 until we get a number between 0 and 25 inclusive. $34-26=8$, meaning that: $$13+21 \equiv 8 \text{ (mod 26)}$$

### Example 2
Calculate $7 \times 5 \text{ (mod 35)}$:

First, we will calculate $7 \times 5$ which will give $35$. An important fact to remember is that $n \equiv 0 \text{ (mod n)}$. This might be a little confusing, but if you think about the clock we saw earlier, it had $12$ numbers on but only went up to $11$ as zero was on the clock instead of the $12$. This means that: $$ 7 \times 5 \equiv 0 \text{ (mod 35)}$$

# Section 3: Shift cipher using numbers

Within this section we will combine what we have learnt in the previous two sections. This will help you to understand cryptography in a more mathematical way. The first step will be to convert our alphabet into numbers, which can be seen in the diagram below:

```python
#| echo: false
#| label: alphabet-with-numbers
from IPython.display import Image, display
display(Image(filename='Alphabet with numbers.png'))
```

As each letter now corresponds to a number, we can introduce a new formula: $$ c = p + s \text{ (mod 26)}$$ where $c$ represents the ciphertext, $p$ represents the plaintext and $s$ represents the shift. Using this formula, if we wanted to encrypt the plaintext letter "r" using a shift ciphper of 5, then we would first see what number "r" corresponds to. We can see that "r" is represented by 17, so to get the ciphertext we do $p + s = 17 + 5 = 22$. We can see that $22$ represents to letter "w", so using a shift cipher of 5 "r" will get encrypted to be "w".

Now we will use these numbers to encrypt the word "Svitlo" with a shift cipher of 3.

### Example 1

To encrypt "Svitlo" with a shift cipher of 3, we first need do convert all the letters into numbers. So we get, $$\text{``Svitlo"}= 18,21,8,19,11,14$$ Then we use the formula and add 3 to these numbers. So we get, $$c=p+s \rightarrow 18+3,21+3,8+3,19+3,11+3,14+3 = 21,24,11,22,14,17$$ Now that we have our ciphertext, we will convert it back into letters to give $$21,24,11,22,14,17 = \text{``Vylwor"}$$

Now that we have looked at that example, you might be thinking *Why have we bothered to turn the letters into numbers? We can easily do a shift cipher without numbers!*, and you would very correct to think this. For a shift cipher it is rather easy to see how the letters will be encrypted, however in future lessons we will look at more complicated ciphers which are not so easy to understand. I have shown you how to turn the letters into numbers with a simple example (shift cipher), so that when we look at harder ciphers you will be able to understand them quicker.

Another reason why we use formulas in encryption is to easily see how to decrypt a message. If we have just recieved the ciphertext from a message, then we can convert it back to the plaintext by rearranging our formula: $$ c = p + s \text{ (mod 26)} \implies p = c - s \text{ (mod 26)}$$

We will now use this idea in an example.

### Example 2
You have recived the ciphertext, "ъсaблвр" which has been created using a shift cipher of 15. What is the plaintext?

Before we start this question, I have to point out the obvious which is that for this example we will be using the Ukrainian alphabet instead of the English alphabet. This means that any calculations we do will have to be in $\text{(mod 33)}$, because the Ukrainian alphabet has 33 letters in. With that in mind, we can start by converting our ciphertext into numbers. We will do this using the following table:

```python
#| echo: false
#| label: ukrainian-alphabet-in-numbers
from IPython.display import Image, display
display(Image(filename='Ukrainian alphabet in numbers.png', width=1000, height=100))
```

From this table, we can see that: $$\text{``ъсaблвр''}=30,21,0,1,15,2,20$$

Now that we have our ciphertext as numbers, we can use the formula $p = c - s \text{ (mod 33)}$ to convert from the ciphertext to the plaintext.

$$30-15,21-15,0-15,1-15,15-15,2-15,20-15 = 15,6,18,19,20,5 \text{ (mod 33)}$$
Now we can convert these numbers (which represent our plaintext) back into letters to find out what the message was.
$$15,6,18,19,20,5=\text{``Леопард''}$$
