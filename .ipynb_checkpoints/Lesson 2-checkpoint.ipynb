# Section 1: Prime factor decomposition

## (Difficulty 2/5)

In Mathematics, you have probably heard of a prime number before. Prime numbers are numbers which can only be divided by one and itself. Examples of prime numbers include: $2,3,5,7,11,13,...$

An example of a number which is not prime would be $15$, because we know that $15= 5 \times 3$. Every number is either a prime number, or is made by multiplying prime numbers together. You can think of prime numbers as being the building blocks of every other number. As prime numbers are so important, then it's useful to be able to find out what prime numbers are used to make up a number. This is called prime factor decomposition, and we will look at an example now.

Lastly, it is important to remember that $1$ is not a prime number. The definition of a prime number is that the number can only be divided by one AND itself. The keyword here is 'and', because this implies that a prime number has to have exactly two factors. However, when we look at the number $1$, we can see that it only has one as a factor, as $1 = 1 \times 1$. As the number $1$ only has one factor, then it is not prime.

### Example 1

Find the prime factor decomposition of $150$. With this example I have created step by step images showing you what to do. I will talk through each of the images so you can follow along and see what is happening at each step.

1. For the first step, all we have to do is simply write the number at the top of the page.
2. In step 2, we have picked two numbers that multiply to give $150$. I have chosen $3$ and $50$.
3. We then look to see if any of these numbers are prime. As $3$ is a prime number, we put a red circle around it to show that it's then end of that branch.
4. We then repeat the process and split 50 into two numbers which multiply to give $50$. I chose $25$ and $2$.
5. As $2$ is prime, we put a red circle around it.
6. We then split $25$ into two numbers which multiply to give $25$. The only two numbers which do that are going to be $5$ and $5$.
7. As 5 is prime, we put red circles around both the number $5$'s.
8. As every branch finishes with a number in a red circle, then our tree is complete. We can write down every number in a red circle and say that they will multiply to give the number at the very top. In this case, $150 = 5^2 \times 2 \times 3$. Note that we have $5^2$ because the number 5 appears twice.

# Section 2: Coprime numbers

## (Difficulty 2/5)

In our previous section, we looked at prime numbers which you have probably heard of before in school. Now I want to introduce you to something that you have probably never seen before, but you will need it to understand our next cipher, so make sure you understand it!

### What are coprime numbers?

Coprime numbers are a pair of numbers which do not share any prime factors with each other. For example, $16$ and $21$ are coprime, because if we look at the prime factor decompositions, then $16=2^4$ and $21=3 \times 7$. As you can see, these do not share any prime factors meaning they are coprime.

However, if we wanted to see if $21$ and $35$ are coprime, then we first need to split them up into their prime factor decomposition. This gives us $21 = 7 \times 3$ and $35 = 7 \times 5$. As you can see, both $21$ and $35$ share the number $7$ as one of their prime factors, meaning that $21$ and $35$ are not coprime.

### Coprime numbers in terms of gcd

Another way of defining a coprime numbers is by saying *'Two numbers are coprime if their greatest common divisor is equal to one'.* Mathematically, we write that two numbers (we can call these numbers $a$ and $b$) are coprime if $gcd(a,b)=1$, where $gcd(a,b)$ means the greatest common divisor of $a$ and $b$.

Now just in case you have never heard of the *greatest common divisor* before, here is the definition. The greatest common divisor of two numbers is the largest number which perfectly divides the two numbers. For example, the greatest common divisor between $24$ and $60$ is equal to $12$, so we would write $gcd(24,60)=12$.

To find the greatest common divisor, we look at the shared prime factors between the two numbers. We can see that the factor trees of $24$ and $60$ are, $24=2^3 \times 3$ and $60=2^2 \times 3 \times 5$. If we were to put these numbers into a Venn diagram, we would get:

````python
#| echo: false
#| label: Venn-diagram-of-factors
from IPython.display import Image, display
display(Image(filename='Venn diagram of factors.png'))
````

We can see that in the middle of the Venn diagram, we have $2, 2, 3$. When we multiply these together, we get the greatest common divisor. We can double check our answer and calculate that $2 \times 2 \times 3 = 12$ just like we expected!

When you use both of these definitions for a coprime number, you will see that they mean the same thing, only they've been said differently. Using the first definition, we can see that two numbers can only be coprime if they do not share any prime factors, and if two numbers do not share any prime factors, then their greatest common divisor must be equal to zero.

Like usual, we are going to have a look at an example before doing any practice questions.

Last week we looked at a shift cipher, where you add a 'shift' onto the plaintext to change it into ciphertext. With an affine cipher, we do two changes to the plaintext. We first multiply the plaintext by some number, then add a shift onto our multiplied plaintext. An affine ciper (for the English alphabet) is defined as follows:

Choose a number $\alpha$ such that $gcd(26,\alpha)=1$ then choose some number $\beta$. The cipher sends each plaintext letter $p$ to ciphertext letter $C= \alpha p + \beta \text{ (mod 26)}$

Time to look at some examples using this cipher.

### Example 1
Encrypt the plaintext letter 'H' using an affine cipher with $\alpha = 5$ and $\beta = 9$.

The first step will be to convert the letter 'H' into a number. Using the alphabet above, we can see that: $$H=7$$

Now we can use our cipher. First we multiply our plaintext by our $\alpha$, then we add our $\beta$. This means we will have: $$5 \times 7 +9 = 35 + 9 = 44 \equiv 18 \text{ (mod 26)}$$

Then to convert $18$ into our ciphertext, we can check the alphabet above. We can see that 'S'=18. This gives us our final answer, so we know that using an affine cipher with $\alpha = 5$ and $\beta = 9$ the plaintext letter 'H' will be encrypted into the ciphertext letter 'S'.

### Example 2

Encrypt the Ukrainian word 'вівця' using an affine cipher with $\alpha = 26$ and $\beta = 3$. Remember that because we are now using the Ukrainian alphabet, all calculations need to be done $\text{mod 33}$.

First, we will convert our word into numbers. We can use the alphabet above to give: $$\text{в}=2, \text{і}=11, \text{ц}=26, \text{я}=32$$

Once we have our numbers, then we can use the cipher. We will get: $$2 \times 26+3=52+3=55 \equiv 22 \text{ (mod 33)}$$ $$11 \times 26+3=286+3=289 \equiv 25 \text{ (mod 33)}$$ $$26 \times 26+3=676+3=679 \equiv 19 \text{ (mod 33)}$$ $$32 \times 26+3=832+3=835 \equiv 7 \text{ (mod 33)}$$

Now we can convert these numbers back into letters. Using the Ukrainian alphabet above we will get: $$22=\text{т},25=\text{х},19=\text{п},10=\text{и}$$



When we encrypt the work 'вівця' using an affine cipher with $\alpha = 26$ and $\beta = 3$ then we get the ciphertext 'тхтпи'.

### Example 3

I have encrypted an English word using an affine cipher with $\alpha = 5$ and $\beta = 10$. The ciphertext is 'QKBER'. What is the English word that I encrypted?

The first step is to convert the ciphertext into numbers. Using the alphabet we can see that: $$Q=16,K=10,B=1,E=4,R=17$$

We know that for an affine cipher: $$\text{Ciphertext} = \alpha \times \text{Plaintext} + \beta \text{ (mod 26)}$$ 

Rearranging this will give: $$ \text{Ciphertext} - \beta = \alpha \times \text{Plaintext} \text{ (mod 26)}$$

We need to be a little bit clever for this next step, as we need to make sure that the value we calculate for the Plaintext is an integer. For the first letter (Q), we will do: $$16-10=5 \times \text{Plaintext} \text{ (mod 26)} \implies 6=5 \times \text{Plaintext} \text{ (mod 26)}$$

We can see here that if we divided both sides of the equation by $5$, then we would not get an integer value for $\text{Plaintext}$. This means we need to keep on adding 26 onto the left hand side of the equation until we get a multiple of $5$. We can add $6$ onto multiples of $26$ until we find a multiple of $5$: $$26+6=32$$ $$52+6=58$$ $$78+6=84$$ $$104+6=110$$

We can see that $110$ is a multiple of $5$, meaning we can write:$$6=5 \times \text{Plaintext} \text{ (mod 26)} \implies 110 = 5 \times \text{Plaintext} \text{ (mod 26)}$$

Now we are able to divide both sides of the equation by $5$, to give:$$110 = 5 \times \text{Plaintext} \text{ (mod 26)} \implies 22 = \text{Plaintext} \text{ (mod 26)}$$

We can see that $22$ corresponds to W. Meaning that the first letter of our plaintext is W. We can now use the same method to find the remaining letters for our plaintext.

For our next letter K: $$10-10=5 \times \text{Plaintext} \text{ (mod 26)} \implies 0=5 \times \text{Plaintext} \text{ (mod 26)}$$

This means that $\text{Plaintext}$ must be equal to zero. We do not need to start adding 26. As $\text{Plaintext}$ is equal to zero, then the next letter in our decoded message must be A.

For our third letter B: $$1-10=5 \times \text{Plaintext} \text{ (mod 26)} \implies -9=5 \times \text{Plaintext} \text{ (mod 26)}$$

We can add multiples of 26 onto the left hand side of this equation to get: $$95=5 \times \text{Plaintext} \text{ (mod 26)}$$

Dividing both sides by 5 we will get:$$95=5 \times \text{Plaintext} \text{ (mod 26)} \implies 19 = \text{Plaintext} \text{ (mod 26)}$$

So our third letter in our decoded work will be T.

For our fourth letter E:

$$4-10=5 \times \text{Plaintext} \text{ (mod 26)} \implies -6=5 \times \text{Plaintext} \text{ (mod 26)}$$

For this equation we can see that we only need to add 26 once to the left hand side to get a multiple of 5. Doing this, we will get: $$-6=5 \times \text{Plaintext} \text{ (mod 26)} \implies 20=5 \times \text{Plaintext} \text{ (mod 26)} \implies 4 = \text{Plaintext} \text{ (mod 26)}$$

This means that our fourth decoded letter is E. Note how for this specific letter, the ciphertext is equal to the plaintext.

For our final letter R:

$$17-10=5 \times \text{Plaintext} \text{ (mod 26)} \implies 7=5 \times \text{Plaintext} \text{ (mod 26)}$$

By adding multiples of 26 to the left hand side of the equation, we will get:$$7=5 \times \text{Plaintext} \text{ (mod 26)} \implies 85=5 \times \text{Plaintext} \text{ (mod 26)} \implies 17 = \text{Plaintext} \text{ (mod 26)}$$

This means that our final decoded letter is R.

Our last step is to put all of our decoded letters together. By doing that, we can see that the decrypted message is WATER.

## Extension work

I am including this section to answer a question which I had when I first learnt about this topic. Please note that this section is not needed to understand any of the content further on in this course, it is only here in case any of you are interested in learning a little bit more Maths. The question I want to answer is:<br>
*'Why does $\alpha$ have to be coprime to the number of letters in the alphabet that we're using?'* <br>

The first thing we need to ask ourselves is, *'Why does a cipher work?'*, and one of the reasons why a cipher works is because it maps each letter from the plaintext onto one unique letter in the ciphertext. In other words, no two letters in the plaintext will map to the same letter in the ciphertext. If this did happen then it would be impossible to decrypt the ciphertext, as you would not know which letter from the plaintext was used.<br> For example, imagine a cipher mapped the letter 'A' and the letter 'G' to the letter 'K'. If I then gave you the letter 'K' and asked you which letter in the plaintext was encrypted to give you 'K', you would have no idea because it could be either 'A' or 'G'.

So, we know that two letters in the ciphertext cannot be equal to each other if they have come from different plaintext letters. Now to answer our question of why $\alpha$ has to be coprime, we are going to use a technique called proof by contradiction. If we want to prove that $\alpha$ has to be coprime with 26 (for the English alphabet), then we need to prove it is impossible to have a valid cipher when $\alpha$ is **not** coprime with 26. I know that this might be confusing if you have never seen proof by contradiction before, so if you haven't, then I would recommend reading about it or asking me in the lesson. Proof by contradiction is a very powerful tool, so if you plan on doing a lot of Maths in the future, then I recommend learning about it now.

Anyway, back to our proof. Imagine we have some affine cipher with an $\alpha$ which is not coprime to 26. We can now choose two values from the plaintext $n$ and $m$ (remember these represent numbers, and not letters). If these did represent the same letters in the ciphertext, then we would get: $$ \alpha n + \beta \equiv \alpha m + \beta \text{ (mod 26)}$$

Immediately we can take away $\beta$ from both sides to get: $$ \alpha n \equiv \alpha m \text{ (mod 26)}$$

If we know that $\alpha$ is coprime with 26, then there will be some value of $\alpha$ which is a factor of 26. There must also be some value of $n$ such that $\alpha n$ is equal to a multiple of 26 (for example, $\alpha$ = 2 and $n$ =13). This means that if $\alpha n$ is equal to a multiple of 26, then we will get: $$\alpha n \equiv \alpha m \text{(mod 26)} \implies 0 \equiv \alpha m \text{ (mod 26)}$$

This equation is true when $n=m$, which is good because if $n=m$ then we have the same letter in the plaintext being mapped to the same letter in the ciphertext. However, this equation is also true for $m=0$, which will break our cipher as it means that two different letters in the plaintext are being mapped to the same letter in the ciphertext. We know that $n$ is not equal to zero as we said earlier that $\alpha n$ is a multiple of 26, which cannot be true if $n$ is zero. Therefore, in order to have a valid affine cipher, $\alpha$ has to be coprime with 26.
