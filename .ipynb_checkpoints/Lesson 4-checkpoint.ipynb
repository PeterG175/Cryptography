Every cipher that you have seen so far has been a mono-alphabetic substitution cipher. This means that one letter in the plaintext represents one letter in the ciphertext, and you just substitute the letters. As we have already seen, this makes them easy to break using a frequency analysis. So this week, we will be investigating ciphers where one letter in the plaintext can represent multiple letters in the ciphertext. We call these ciphers poly-alphabetic ciphers.

# Section 1: Vigenere cipher

Note that the second 'e' in Vigenere has an accent on it, but I do not know how to write that on this website.

The Vigenere cipher was first described in 1553 by Giovan Battista Bellaso in *La cifra del. Sig. Giovan Battista Bellaso* but has mistakenly been named after Blaise de Vigenere who described something similar in *Traite des Chiffres* in 1586. This cipher was not broken for 300 years because of how strong it is.

In our other ciphers, our key has been a number (for a shift cipher) and two numbers (for our affine cipher). In a Vigenere cipher, our key is a word. This is sensibly called our 'keyword'. The cipher works by converting our keyword into numbers, then calculating: $$ c = p + k \text{ (mod 26)}$$

where $c$ is the ciphertext, $p$ is the plaintext and $k$ is the keyword.

Once you get to the end of the keyword, you then start it again over the next set of plaintext letters and repeat the process. We will now see how this works in an example.

Our example is: encrypt the message 'Do you want to play chess' using a Vigenere cipher with the keyword 'Horse'.

Now that we have our plaintext in number form, we can convert our key into numbers. Doing this we get:

$$\text{Horse}=7,14,17,18,4$$

Next we will put the key next to the plaintext in the table.

```python
#| echo: false
#| label: Vigenere-cipher-example-12
from IPython.display import Image, display
display(Image(filename='Vigenere cipher example 12.png'))
```

To get the ciphertext, we need to add the plaintext to the key. As we already have all the numbers in this table, then it becomes quite easy to add the numbers together. Just remember that all the calculations are done $\text{mod 26}$. So for example, in the 3rd column, we have $24+17$ which is $41$, however as this is greater than $26$ we need to take $26$ away from $41$. Hence $24+17=41 \equiv 15 \text{ (mod 26)}$.

Adding up all the numbers, we get:

```python
#| echo: false
#| label: Vigenere-cipher-example-13
from IPython.display import Image, display
display(Image(filename='Vigenere cipher example 13.png'))
```

Now our final step is to convert these numbers back into letters. This will give us our final table:

```python
#| echo: false
#| label: Vigenere-cipher-example-14
from IPython.display import Image, display
display(Image(filename='Vigenere cipher example 14.png'))
```

And that gives us our ciphertext!

So to answer the question, encrypting the message '$\text{Do you want to play chess}$' using a Vigenere cipher with the keyword as 'Horse', will give the ciphertext: <br>
'$\text{KCPGYDOFLXVDCSCJVVKW}$'.

Before we look at another example, I think it's a good idea to pause for a moment to look at what's happened to each letter in the previous example. We can see that in the plaintext, we have a letter 'O' in the 2nd and 4th columns. In a shift or affine cipher, these letters would be encrypted to the same ciphertext letter. However, as we can see, the first letter 'O' gets encrypted to the ciphertext letter 'C' and the second 'O' gets encrypted to the ciphertext letter 'G'. The same plaintext letters get mapped to different ciphertext letters, which means a standard frequency analysis will not be able to decrypt the ciphertext. <br>
More good news is that it is still easy to decrpyt the ciphertext if you know the keyword, all you have to do is subtract the keyword from the ciphertext to get back to the plaintext.

# Section 2: Autokey cipher

An autokey cipher is another cipher which is stronger that a simple substitution cipher. This cipher uses a *primer* key which is put infront of the plaintext to encrypt the first few plaintext letters, then the plaintext is used to ecrypt further letters! This sounds unusual but it should make sense using an example.

### Example 1

The first step is to convert our message from letters into numbers:

Now we put our primer key in the first part of the message, and then use the plaintext as our encryption key.

```python
#| echo: false
#| label: Autokey-cipher-example-12
from IPython.display import Image, display
display(Image(filename='Autokey cipher example 12.png'))
```

We can then convert our key into numbers and add the plaintext to the key to give us the ciphertext.

```python
#| echo: false
#| label: Autokey-cipher-example-13
from IPython.display import Image, display
display(Image(filename='Autokey cipher example 13.png'))
```

So encrypting the message 'Cryptography is fun' using an autokey cipher with primer 'Study' will give the ciphertext 'UKSSRQXPPADEZSUBL'.

### Example 2

I have ecrypted a secret message using an autokey cipher with the primer being 'Scooby'. The ciphertext is 'AYWZMNTWGHTJSMWZWZ', what is the plaintext?


This encryption takes a bit longer to break than other ciphers that we've seen so far. There is probably a quicker way to solve it than the method I am going to show you, however I think this method is quite intuitive. We will have to break down our ciphertext into blocks and then decrypt each block seperately.

As the primer is 6 letters long, we will break the first 6 letters of the ciphertext to begin with. As usual, we convert the key and ciphertext into numbers.

```python
#| echo: false
#| label: Autokey-cipher-example-21
from IPython.display import Image, display
display(Image(filename='Autokey cipher example 21.png'))
```

As we are decrypting the message (and not encrypting it), we will have to take the primer away from the ciphertext. Remember that all the calculations are done with $\text{mod 26}$ This will give us the following:

```python
#| echo: false
#| label: Autokey-cipher-example-22
from IPython.display import Image, display
display(Image(filename='Autokey cipher example 22.png'))
```

Then we will convert this back into text:

```python
#| echo: false
#| label: Autokey-cipher-example-23
from IPython.display import Image, display
display(Image(filename='Autokey cipher example 23.png'))
```

Now that we have our first part of the plaintext, we need to use this to help decrypt the next block of our ciphertext.

Our next block of ciphertext will also be 6 letters long, which means it is 'TWGHTJ'. We will convert that part of the ciphertext into numbers, then take our first block of plaintext away from it. Then by converting the answer back into letters, we get this table:

```python
#| echo: false
#| label: Autokey-cipher-example-24
from IPython.display import Image, display
display(Image(filename='Autokey cipher example 24.png'))
```

We only have 6 more letters of our ciphertext to decrypt before we have our full message.

We will use the same method as we used for the previous block.

First, we convert the last block of ciphertext into numbers, this block is 'SMWZWZ'. Then we take away the second part of the plaintext from it, this part is 'laywit'. Then when we get our answer, we will convert it back into letters and we'll have decrypted the final part of our ciphertext.

Putting all of this into a table, we get:

```python
#| echo: false
#| label: Autokey-cipher-example-25
from IPython.display import Image, display
display(Image(filename='Autokey cipher example 25.png'))
```
